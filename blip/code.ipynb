```python
from transformers import BlipProcessor, BlipForConditionalGeneration
from PIL import Image
import torch, os
import matplotlib.pyplot as plt

# Device setup
device = "cuda" if torch.cuda.is_available() else "cpu"

# Load BLIP
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base").to(device)

# Your cropped objects folder
save_folder = "/scratch/codepk37/anlpnerf/sam2/notebooks/mask_obj_colormugs_take2"

pri=[]

# Loop through cropped objects
for file in sorted(os.listdir(save_folder)):
    if file.endswith(".png"):
        img_path = os.path.join(save_folder, file)
        img = Image.open(img_path).convert("RGB")

        # Generate caption
        inputs = processor(img, text="a photo of", return_tensors="pt").to(device)
        out = model.generate(**inputs, max_new_tokens=10)
        caption = processor.decode(out[0], skip_special_tokens=True)

        # Show image + caption
        plt.figure(figsize=(2,2))
        plt.imshow(img)
        plt.axis("off")
        plt.title(caption, fontsize=10)
        plt.show()
        pri.append(caption)

print(pri)
```
